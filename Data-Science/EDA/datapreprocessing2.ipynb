```python
import pandas as pd 
import numpy as np


# example dataset 
numeric_data = pd.DataFrame({
    'Salary': [30000,45000,60000,800000,120000],
    'Age': [25,42,30,51,55]
})
numeric_data
print(numeric_data)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
#dtandard scaler (z socre )
std_scaler = StandardScaler()
scaled_std = std_scaler.fit_transform(numeric_data)
df=pd.DataFrame(scaled_std, columns=numeric_data.columns)
print("\nStandard Scaled Data:\n",df)
```

fit_transform does two things 
fit(): learn the mean and standard s=deviation of ur data 
transform (): Applies  the scaling (converts data into z-sscore)
(for stnadardscaler: Each feature became mean=0 and std = 1)


```python
minmax_scaler = MinMaxScaler()
scaled_minmax = minmax_scaler.fit_transform(numeric_data)
print("\nMin-Max Scaled Data:\n",pd.DataFrame(scaled_minmax, columns=numeric_data.columns))
```

```python
# Encoding categorical value 
# encoding works categorlical , aphalbetical - hr , marketing - A,B , c  

from sklearn.preprocessing import LabelEncoder,OneHotEncoder
# sample data 
cat_data = pd.DataFrame({
    'Department': [ 'HR', 'Sales', 'IT', 'HR','IT']
})
cat_data
```

```python
#Labelencoder 

label_encoder= LabelEncoder()
cat_data['Dept_Label']= label_encoder.fit_transform(cat_data['Department'])
print("\nLabel Encoder :\n", cat_data)
```
