```python
import numpy as np
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
# craeting a sample dataframe  for distribution
data={
    'Income':[25000,35000,45000,60000,80000],
    'Car_Color':['Red','Red','Black','Blue','Beige']
}
df = pd.DataFrame(data )
df
```

```python
mean_income = df['Income'].mean()
print("Mean Income", mean_income)
```

```python
median_income = df['Income'].median()
print("Median Income", median_income)
```

```python
mode_color = df['Car_Color'].mode()[0]
print("Most Common Color",mode_color)
```

```python
std_income = df['Income'].std()
var_income = df['Income'].var() 
print(" Standard Deveiation:", std_income)
print(" Variance:", var_income )
```

```python
range_income = df['Income'].max()-df['Income'].min()
print( " Range Of Income :", range_income )
```
