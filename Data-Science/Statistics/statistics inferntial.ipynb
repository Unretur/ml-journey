```python
import numpy as np
population = np.arange(1,10001) # 10000 users
sample = np.random.choice(population, size= 500 ,replace = False )
print(sample[:5])
```

```python
mean=[]
for i in range(1000):
    sample = np.random.choice(population , 100)

print(sample[:9])
```

```python
mean.append(np.mean(sample))

import matplotlib.pyplot as plt
plt.hist(mean,bins=30)
plt.title('sample distribution by sample mean')
plt.show()
```
