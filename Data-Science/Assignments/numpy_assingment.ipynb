```python
import numpy as np
import time 
```

```python
arr=np.arange(1,21)
print(arr)
```

```python
print("\nSum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
```

```python
print("\nIndices of elements greater than 10:")
print(np.where(arr > 10))
```
