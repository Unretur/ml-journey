```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
time = np.arange(1,101)
non_stationary = 50+0.5*time + np.random.normal(size=100)
non_stationary
```

```python
# statinary series
stationary = np.random.normal(loc=50,scale=5,size=100)
stationary
```

```python
#plot bith od them 
plt.figure(figsize=(12,15))
plt.subplot(1,2,1)
plt.plot(time,non_stationary, label ="non stationary(Trend)", color='red')
plt.title("Non-sationary series")
plt.xlabel("Time")
plt.ylabel("value")
plt.legend()
plt.grid(True)
 
plt.subplot(1,2,1)
plt.plot(time,stationary, label ="stationary(Random Noise)", color='green')
plt.title("sationary series")
plt.xlabel("Time")
plt.ylabel("value")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
## aug (augmnented dickey filler)

from statsmodels.tsa.stattools import adfuller

result = adfuller(non_stationary)
print("ADF statistic", result[0])
print("p-value", result[1])
```

```python
result1 = adfuller(stationary)
print("ADF statistic", result[0])
print("p-value", result[1])
```

```python
# first differnceing 
diff_series = pd.Series(non_stationary).diff().dropna()
diff_series.head()
```

```python
#plot them

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(non_stationary, label ="original (non-stationary)", color='red')
plt.legend()
plt.grid(True)

plt.figure(figsize=(12,15))
plt.subplot(1,2,1)
plt.plot(diff_series, label ="after difference stationary", color='blue')
plt.legend()
plt.grid(True)
```

```python
# adf result
result_diff = adfuller(diff_series)
print("after difference ", result_diff[1])
```
