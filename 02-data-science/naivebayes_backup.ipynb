```python
## import important libaries 

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report 
from sklearn.naive_bayes import GaussianNB
from sklearn.impute import SimpleImputer
```

```python
## load the datas of titanic dataset 


df = pd.read_csv("titanic.csv")
df.head()
```

```python
#  3 sleect and prepare featutre 
df_model = df[["Survived","Pclass","Sex","Age","Fare"]]

# convert sex to numeric terms 
df_model["Sex"] = df_model["Sex"].map({"male": 0, "female": 1})
```

```python
df_model.head()
```

```python
df_model.isnull().sum()
```

```python
df_model.shape
```

```python
# step 4 handel misiing data 

# sperate feature x and atrget  y
X = df_model.drop("Survived", axis=1) 
y = df_model["Survived"]

# use simpleimputer to fill missing age with mean 
imputer = SimpleImputer(strategy='mean') # mean imputattion for numerical columns
X_imputer = imputer.fit_transform(X)

```

```python
X.head()
```

```python
y.head()
```

```python
# now x is impute with no missing value 

import numpy as np 
np.isnan(X_imputer).any()
```

```python
 # train and test spilt 

X_train,X_test,y_train,y_test = train_test_split(X_imputer,y, test_size = 0.2,random_state= 42)
```

```python
# putting naive bayes model intp working 

nb_model = GaussianNB()

# tarin on the training data
nb_model.fit(X_train,y_train)
```

```python
X_test
```

```python
# amke prediction and evaluate 

y_pred = nb_model.predict(X_test)

# evaluation matrix
print("Accuracy :", accuracy_score(y_test,y_pred))
print("\n confusion matrix:\n", confusion_matrix(y_test,y_pred))
print("\n CLassification reprt :\n", classification_report(y_test,y_pred))
```
