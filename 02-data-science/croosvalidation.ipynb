```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LogisticRegression 
```

```python
iris = load_iris()   # load dataset into memory 
X,y = iris.data, iris.target    # fetaure  and target arrays 
```

```python
model = LogisticRegression(max_iter=200, n_jobs= None)

# maximum number of iteration   the solver will able to run in converge 
                 # sometimes LOgistics regression needs more iterations for convergence ,
                 # especially with sclaed or high


cv = KFold(n_splits = 5, shuffle = True,random_state= 42)

scores = cross_val_score(model, X, y , cv =cv , scoring ="accuracy")
```

```python
print("Fold Acuuries :", scores)

print(" Mean Acuuries :", scores.mean())

print("std dev  :", scores.std())

#print(" approx 95% CI : [{:.3f},{:.3f}]".format ( scores.mean()- 1.96*scores.std()/np.sqrt(len(scores))), 
                                                   # scores.mean()+ 1.96*scores.std()/np.sqrt(len(scores)))
```
