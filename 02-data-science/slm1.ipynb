```python
## defien idependent ( footfall  ) and dependent (daily sales ) variables

X= data[['Footfalls']]
Y= data['DailySales']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X,Y)
```

```python
## PRINT THE model parameters

print (f"Coefficent (slope): {model.coef_}")
print(f"Interecpt: {model.intercept_} ")
```

simple linear regression equation = y =B1+B0*x

where :
y =  predicted value (dependent variable , eg ,- sales)
x = input va;ue (independent variable , eg , - footfall)
B1 = slope ( coefficent showing how much y changes each x)
B0 = interecpt (value of x )
