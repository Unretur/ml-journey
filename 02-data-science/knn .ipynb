```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split
from collections import Counter
```

```python
columns_names = [
    "Area","perimeter","Compactness","KernelLength","kernalWeight",
    "AsymmetryCoeff","GrooveLength","Class"
]

data = pd.read_csv('seeds_dataset.txt', delim_whitespace= True, names=columns_names)
data.head()
```

```python
## step  3 - prepare feature and labels

X = data.iloc[:,:-1].values   # all columns except last 
Y =data.iloc[:, -1].values  # lastt column = class labels (1,2,3)
```

```python
## step  4 - train and test spliy
X_train,X_test,Y_train,Y_test = train_test_split(X,Y, test_size = 0.2, random_state=1)
```

```python
# step 5 define eulcidean distance function
def euclidean_distance(X1,X2):
    return np.sqrt(np.sum(X1-X2**2))
```

```python
## step 6 : knn classifeier from scrath

class KNN:
    def _init_(self, k= 3):
        self.k = k
    
    def fit(self , X, Y ):
        self.X_train = X
        self.Y_train = Y 

    def predict(self,X):
        return np.array([self._predict(X)for X in X])   # predict for each sampple

    def _predict(self,X):
        # calucalte distance between X and Y all training samople
        distances = [euclidean_distance(X,X_train) for X_train in self.X_train]

        # find the indices of k smallesr    distamces 
        k_indices = np.argsort(distances)[:self.k]  # np.argsort returns sorted indices

        # collect labesl of the k value 
        K_nearest_labels = [self.Y_train[i] for i in k_indices]

        # count teh frequency of each label 
        most_common = Counter(K_nearest_labels).most_common(1)

        ## rerurn the label with highes count 
        return most_common[0][0]
```

```python
# step 7 : initalized model, train and predict
knn = KNN( K=5)
knn.fit(X_train,Y_train)
Y_pred = knn.predict(X_test)


## step 8 : evaluate model accuracy
accuracy= np.mean(Y_pred == Y_test)
print(f"Accuracy on Seed Dataset :{accuracy:.f}")
```
