```python
label_encoders={}
for column in churn_df.select_dtypes(include=['object']).columns:
    if column !='customerID':
        label_encoders[column]= LabelEncoder()
        churn_df[column] = label_encoders[column].fit_transform(churn_df[column])
        
```

```python
churn_df.head()
```

```python
X = churn_df.drop(columns = ['customerID', 'Churn'])
Y = churn_df['Churn']
X.head()
```

```python
Y.head()
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y, test_size =0.2, random_state=42 )
```

```python
X_test.shape
```

```python
Y_test.shape
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression

# model 
model = LogisticRegression()

#train the model 
model.fit(X_train,Y_train)

#c make prediction

Y_pred = model.predict(X_test)
```

```python
# recursive feature elimination 

from sklearn.feature_selection import RFE

rfe_model = LogisticRegression()

#initation rfe model 
rfe = RFE(estimator= rfe_model, n_features_to_select = 5 )
```

```python
rfe.fit(X_train,Y_train)

# get selected feature

selected_features = rfe.support_

# tarnsform the data 
X_train_rfe = rfe.transform(X_train)
X_test_rfe = rfe.transform(X_test)

# prediction 

Y_pred_rfe = model.predict(X_test_rfe)
```

```python
from sklearn.metrics import confusion_matrix , accuracy_score

# gerante the confusion matrix
conff_matrix = confusion_matrix(Y_test, Y_pred_rfe)
print("confusion Matric :\n" , conff_matrix)
```
