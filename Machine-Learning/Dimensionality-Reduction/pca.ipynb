```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
```

```python
# load dataset (its dataset . 150 folwer , 3species , 4 features)
iris = load_iris()
X = iris.data   # feature : sepal lenght, sepal wight , petal lenght and weight
y = iris.target # sepecies of iris

target_names = iris.target_names
X[:5]
```

```python
target_names
```

```python
X_scaled = StandardScaler().fit_transform(X)
```

```python
pca = PCA(n_components=4)
X_pca = pca.fit_transform(X_scaled)

X_pca[:5]
```

```python
print("Explained variance ratio :", pca.explained_variance_ratio_)
```

```python
print("Cumulative varainace explained :", np.cumsum(pca.explained_variance_ratio_))
```
