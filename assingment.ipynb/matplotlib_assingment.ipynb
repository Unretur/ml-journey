```python
import matplotlib.pyplot as plt
```

```python
# plot the data on line garph 
x = [1,2,3,4,5]
y = [10,15,25,30,50]

plt.plot(x,y)
#b. chainging tittland grid and customized 
plt.title("Line Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()

plt.show()
```

#2 question
craeting a bar garph where customizing with changes 

```python
#a plot the garph 
students = ["John","Jane","Alice","Bob"]
marks = [75,85,60,90]

#b customized with colour and tittle

plt.bar(students,marks,color=["red","green","blue","orange"])

plt.title("Students Marks")

plt.show()
```

#3 question 
#plot pie garph 

```python
regions = ["North America","Europe","Asia","Others"]

revenue = [45,25,20,10]

# b highletd region with highest revnue 

plt.pie(
    revenue,
    labels=regions,
    explode=(0.1,0,0,0),
    autopct="%1.1f%%"
)

plt.title("Company Revenue")

plt.show()
```

#4 question
#import numpy for random int from 1 to 1000 in histogram


```python
import numpy as np

data = np.random.randint(1,101,1000)

plt.hist(data,bins=20)

plt.title("Histogram")

plt.xlabel("Values")

plt.ylabel("Frequency")

plt.show()
```
