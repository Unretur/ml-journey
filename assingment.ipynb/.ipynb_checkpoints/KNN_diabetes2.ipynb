```python
# Euclidean Distance Function
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))


# Custom KNN Class
class KNN:

    def __init__(self, k=5):
        self.k = k

    # Store training data
    def fit(self, X, y):
        self.X_train = np.array(X)
        self.y_train = np.array(y)

    # Predict multiple samples
    def predict(self, X):
        predictions = [self._predict(x) for x in X]
        return np.array(predictions)

    # Predict one sample
    def _predict(self, x):

        # Calculate distance from every training sample
        distances = [euclidean_distance(x, x_train) for x_train in self.X_train]

        # Get indices of k nearest neighbors
        k_indices = np.argsort(distances)[:self.k]

        # Get labels of nearest neighbors
        k_nearest_labels = [self.y_train[i] for i in k_indices]

        # Majority voting
        most_common = Counter(k_nearest_labels).most_common(1)

        return most_common[0][0]
```

now we will start predicting the test data and like match it with actual data 

```python
knn = KNN(k=5)

knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
```

now we will start with accuracy we have and our model have performe

```python
## Accuracy of knn model
print("Accuracy:", accuracy_score(y_test, y_pred))
```

the accuracy is only 69 percent by all the method and way 
this means the model is predicting 69 correct out of 100 paitents 

```python
 ## now we will move forward with classification report 

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

```python
# confusion matrix 

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))
```
