# 1 basic of probality

the coin will be tossed 10000 times we need to calucate the orobality of number of heads and tails . as number of expriment increase the probality approach towards 0.5

```python
import random

tosses = 10000

heads = 0
tails = 0

for i in range(tosses):
    result = random.choice(["H", "T"])

    if result == "H":
        heads += 1
    else:
        tails += 1

print("Heads:", heads)
print("Tails:", tails)

print("Experimental Probability of Heads =", heads/tosses)
print("Experimental Probability of Tails =", tails/tosses)
```

# 1b . rolling two dice 
the dice will be rolled 10000 time . we need to caluclate the probality of how many times the sum will equal to zero

```python
trials = 10000
count = 0

for i in range(trials):

    die1 = random.randint(1,6)
    die2 = random.randint(1,6)

    if die1 + die2 == 7:
        count += 1

print("Number of times sum = 7:", count)
print("Experimental Probability =", count/trials)
```

```python
#3 condtional probality of bayes theorm 
bag = ["Red"]*5 + ["Green"]*7 + ["Blue"]*8

draws = []

for i in range(1000):
    draws.append(random.choice(bag))

blue_then_red = 0
blue_previous = 0

for i in range(1, len(draws)):

    if draws[i-1] == "Blue":

        blue_previous += 1

        if draws[i] == "Red":
            blue_then_red += 1

conditional_probability = blue_then_red / blue_previous

print("P(Red | Previous Blue) =", conditional_probability)

# Bayes Theorem Verification

P_red = draws.count("Red") / len(draws)
P_blue = draws.count("Blue") / len(draws)

P_blue_given_red = P_blue

bayes = (P_blue_given_red * P_red) / P_blue

print("Bayes Theorem Probability =", bayes)
```

4 . RANDOM variable 
if a 1000 random observation is genrated from probality distribution . we need to calucalte emprical mean , variance and standard deviation

```python
import numpy as np

values = [1,2,3]

probabilities = [0.25,0.35,0.40]

sample = np.random.choice(values,
                          size=1000,
                          p=probabilities)

print("Mean =", np.mean(sample))
print("Variance =", np.var(sample))
print("Standard Deviation =", np.std(sample))
```

5 . continuous random variables 
expontential probality with mean 5 . histogram plotted along with theoritical probality of density function 

```python
import matplotlib.pyplot as plt

sample = np.random.exponential(scale=5,size=2000)

plt.figure(figsize=(8,5))

plt.hist(sample,
         bins=30,
         density=True,
         alpha=0.6,
         label="Histogram")

x = np.linspace(0,max(sample),200)

pdf = (1/5)*np.exp(-x/5)

plt.plot(x,pdf,linewidth=2,label="PDF")

plt.title("Exponential Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()

plt.show()
```

```python
# 6 . central limit theorem 

population = np.random.uniform(0,1,10000)

sample_means = []

for i in range(1000):

    sample = np.random.choice(population,
                              size=30)

    sample_means.append(np.mean(sample))

plt.figure(figsize=(8,5))

plt.hist(population,
         bins=30,
         alpha=0.6,
         density=True,
         label="Uniform Distribution")

plt.title("Uniform Distribution")
plt.legend()

plt.show()

plt.figure(figsize=(8,5))

plt.hist(sample_means,
         bins=30,
         density=True)

plt.title("Distribution of Sample Means")

plt.show()
```
