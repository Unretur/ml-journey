```python
import pandas as pd
```

```python

data = {
    'Name':['Alice','Bob','Charlie','Diana','Eve'],
    'Age':[24,27,22,32,29],
    'Department':['HR','Finance','IT','Marketing','HR'],
    'Salary':[45000,54000,50000,62000,47000]
}
df = pd.DataFrame(data)
```

```python
#a getting data frame 
print(df.head())
```

```python
#b summary for age and salary 
print(df[['Age','Salary']].describe())
```

```python
#c. getting average salary of any department . in this acse according to question of h.r department 
avg_salary = df[df['Department']=='HR']['Salary'].mean()

print("Average HR Salary:",avg_salary)
```

```python
df['Bonus'] = df['Salary']*0.10

print(df)
```

#3 question
#for gettting specific age of the employee in the firm . in this case 25-30.

```python
filtered = df[(df['Age']>=25) & (df['Age']<=30)]

print(filtered)
```

#4 question 
#grouping data by depatmnet and their salry by functuion groupby 

```python
group = df.groupby('Department')['Salary'].mean()

print(group)
```

```python
sorted_df = df.sort_values(by='Salary')

print(sorted_df)

sorted_df.to_csv("sorted_employee_data.csv",index=False)
```
